#1. Grocery Billing

```python
p1=int(input('Enter price of first item '))
p2=int(input('Enter price of second item '))
i1=int(input('Enter quantity of first item '))
i2=int(input('Enter quantity of second item '))
total=p1*i1+p2*i2
print('Total Bill =  ',total)
```

#2. Temperature Conversion

#3. Salary Slip

```python
s=int(input('Enter salary '))
hra=0.20*s
da=0.10*s
ns=s+hra+da
print('Net Salary = ',ns)
```

#4. Simple Interest

```python
p=(int(input('Enter principal amount ')))
r=(int(input('Enter rate of interest ')))
t=(int(input('Enter time ')))
si=(p*r*t)/100
print('Simple Interest = ',si)
```

#7. Discount Calculation

```python
x=int(input('Enter The Original Price : '))
d=(x/100)*10
dp=x-d
print('Discounted Price = ',dp)
```

#8. Area Of Triangle (Using Heron's Formula)

```python
from math import sqrt
a=int(input('Enter side a : '))
b=int(input('Enter side b : '))
c=int(input('Enter side c : '))
s=(a+b+c)/2
area=sqrt(s*(s-a)*(s-b)*(s-c))
print('Area of Triangle = ',area)
```

#9. Paint Cost Estimator

```python
l=int(input('Enter wall length : '))
w=int(input('Enter wall width : '))
win=(1.5*1.2)*2
a=(l*w)-win
pc=a*25
print('Total area to be painted = ',a,'sq.m')
print('Total painting cost : ',pc)
```
