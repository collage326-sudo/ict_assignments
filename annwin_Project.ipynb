6. Quiz Engine with Dynamic Scoring
Create a Quiz Engine that dynamically evaluates participants.
The system should manage a question bank categorized by difficulty levels. During quiz
attempts, questions should appear in a pseudo-random order without using random
libraries.
Scoring must vary by difficulty, include negative marking, and generate detailed score
reports. String handling and nested data structures must be used extensively

```python
questions = {
    "easy": [
        {"question": "What is 2 + 2?", "options": ["1", "2", "3", "4"], "answer": "4"},
        {"question": "What color is the sky?", "options": ["Blue", "Green", "Red", "Yellow"], "answer": "Blue"},
    ],
    "medium": [
        {"question": "What is 12 * 12?", "options": ["124", "144", "154", "134"], "answer": "144"},
        {"question": "Which planet is known as the Red Planet?", "options": ["Earth", "Mars", "Venus", "Jupiter"], "answer": "Mars"},
    ],
    "hard": [
        {"question": "What is the square root of 256?", "options": ["14", "15", "16", "18"], "answer": "16"},
        {"question": "Who wrote 'Macbeth'?", "options": ["Shakespeare", "Dickens", "Hemingway", "Tolkien"], "answer": "Shakespeare"},
    ]
}

# -------------------------------
# SCORING SYSTEM
# -------------------------------
points = {"easy": 1, "medium": 2, "hard": 3}
negative_mark = -0.5  # for wrong answers

# -------------------------------
# PSEUDO-RANDOM ORDER FUNCTION
# -------------------------------
def pseudo_random_index(length, seed):
    # Simple linear congruential generator (LCG)
    return (seed * 7 + 3) % length

# -------------------------------
# QUIZ FUNCTION
# -------------------------------
def start_quiz():
    print("Welcome to the Quiz!\n")
    total_score = 0
    report = []  # store each question result
    seed = 1     # initial seed for pseudo-random ordering

    # Loop over difficulty levels
    for level in ["easy", "medium", "hard"]:
        qlist = questions[level]
        used = []
        print(f"\n--- {level.upper()} QUESTIONS ---")

        for i in range(len(qlist)):
            # Generate pseudo-random index
            idx = pseudo_random_index(len(qlist), seed)
            while idx in used:
                seed += 1
                idx = pseudo_random_index(len(qlist), seed)
            used.append(idx)
            seed += 1

            q = qlist[idx]
            print(f"\nQ{i+1}: {q['question']}")
            for j, opt in enumerate(q["options"], start=1):
                print(f"{j}. {opt}")

            # Take input
            answer = input("Your answer (type the option text): ").strip()

            # Evaluate
            if answer.lower() == q["answer"].lower():
                score = points[level]
                print(f"Correct! +{score} points.")
            else:
                score = negative_mark
                print(f"Wrong! {negative_mark} points. Correct answer: {q['answer']}")

            total_score += score
            report.append({
                "question": q["question"],
                "your_answer": answer,
                "correct_answer": q["answer"],
                "score": score
            })

    # -------------------------------
    # DETAILED REPORT
    # -------------------------------
    print("\n\n--- QUIZ COMPLETE ---")
    print(f"Total Score: {total_score}\n")
    print("Detailed Report:")
    for r in report:
        print(f"Q: {r['question']}")
        print(f"Your answer: {r['your_answer']}")
        print(f"Correct answer: {r['correct_answer']}")
        print(f"Score: {r['score']}\n")
```

```python

start_quiz()
```

7. Personal Expense Tracker
Develop a Personal Expense Tracker to monitor and analyze spending behavior.
Users should be able to log expenses with date, category, and amount. The system should
generate monthly summaries, category-wise spending reports, budget breach alerts, and
identify high-spending patterns.
Analytical insights like highest expense day and dominant spending category must be
computed using loops and conditionals.


```python
# -------------------------------
# Personal Expense Tracker
# -------------------------------

# Initialize empty expense list
expenses = []

# Default monthly budget
monthly_budget = 1000  # You can change this

# -------------------------------
# Function to add an expense
# -------------------------------
def log_expense():
    print("\nLog a new expense:")
    date = input("Enter date (YYYY-MM-DD): ").strip()
    category = input("Enter category (Food, Transport, Bills, Entertainment, Other): ").strip()
    try:
        amount = float(input("Enter amount: ").strip())
    except:
        print("Invalid amount. Please enter a number.")
        return
    expenses.append({"date": date, "category": category, "amount": amount})
    print(f"Expense logged: {date} | {category} | {amount}")

# -------------------------------
# Function to generate monthly summary
# -------------------------------
def monthly_summary():
    month = input("\nEnter month to summarize (YYYY-MM): ").strip()
    total = 0
    category_totals = {}
    day_totals = {}

    for exp in expenses:
        if exp["date"].startswith(month):
            # Total spending
            total += exp["amount"]

            # Category-wise spending
            cat = exp["category"]
            if cat not in category_totals:
                category_totals[cat] = 0
            category_totals[cat] += exp["amount"]

            # Day-wise spending
            day = exp["date"]
            if day not in day_totals:
                day_totals[day] = 0
            day_totals[day] += exp["amount"]

    print(f"\n--- Monthly Summary for {month} ---")
    print(f"Total Spending: {total}")

    # Category-wise report
    print("\nCategory-wise spending:")
    for cat, amt in category_totals.items():
        print(f"{cat}: {amt}")

    # Budget breach alert
    if total > monthly_budget:
        print("\n⚠ Budget exceeded!")

    # Highest expense day
    if day_totals:
        max_day = max(day_totals, key=day_totals.get)
        print(f"Highest expense day: {max_day} ({day_totals[max_day]})")

    # Dominant spending category
    if category_totals:
        dominant_cat = max(category_totals, key=category_totals.get)
        print(f"Dominant spending category: {dominant_cat} ({category_totals[dominant_cat]})")

# -------------------------------
# Function to show all expenses
# -------------------------------
def show_expenses():
    if not expenses:
        print("\nNo expenses logged yet.")
        return
    print("\n--- All Expenses ---")
    for i, exp in enumerate(expenses, start=1):
        print(f"{i}. {exp['date']} | {exp['category']} | {exp['amount']}")

# -------------------------------
# Main Menu
# -------------------------------
def main():
    while True:
        print("\n--- Personal Expense Tracker ---")
        print("1. Log Expense")
        print("2. Show All Expenses")
        print("3. Monthly Summary")
        print("4. Exit")
        choice = input("Enter your choice: ").strip()

        if choice == "1":
            log_expense()
        elif choice == "2":
            show_expenses()
        elif choice == "3":
            monthly_summary()
        elif choice == "4":
            print("Exiting tracker. Goodbye!")
            break
        else:
            print("Invalid choice. Please try again.")
```

```python
# Run the tracker
main()
```
